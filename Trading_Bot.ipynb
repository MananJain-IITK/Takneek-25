```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import math
import seaborn as sns
import statsmodels.api as sm
import plotly.graph_objects as go
from dataclasses import dataclass
from typing import Optional, Tuple
from scipy.optimize import minimize
from warnings import filterwarnings
filterwarnings('ignore')


import pandas as pd
import plotly.express as px

def plot_correlation_heatmap(assets):
    # Collect ask_px1 from each asset
    data = {}
    for idx, asset in enumerate(assets, start=1):
        if "ask_px1" in asset.columns:
            data[f"Asset {idx}"] = asset["ask_px1"].values

    # Build DataFrame (aligns columns by index length)
    df = pd.DataFrame(data)

    # Compute correlation
    corr_matrix = df.corr()

    # Plot heatmap
    fig = px.imshow(
        corr_matrix,
        text_auto=".2f",
        color_continuous_scale="RdBu_r",
        title="Correlation Heatmap of ask_px1 across Assets"
    )
    fig.update_layout(
        xaxis_title="Assets",
        yaxis_title="Assets",
        width=700,
        height=600
    )
    fig.show()


#Strategy_Classes/Functions
#Exploiting relation between Asset1 and NoTrade
import numpy as np
from scipy.signal import correlate

def analyze_signal_correlation(obs, notrade):
    # Extract and clean trading signal
    signal = notrade['signal'].dropna().reset_index(drop=True)

    def preprocess(series):
        scaled = series.diff().diff().diff()
        return scaled.dropna().reset_index(drop=True)

    results = []

    for k, df in enumerate(obs):
        ask_series = preprocess(df['ask_px1'])

        # Align lengths
        min_len = min(len(signal), len(ask_series))
        sig = signal[:min_len]
        ask = ask_series[:min_len]

        # Cross-correlation
        xcorr = correlate(sig - sig.mean(), ask - ask.mean(), mode="full")
        lags = np.arange(-len(sig) + 1, len(sig))

        best_idx = np.argmax(np.abs(xcorr))
        best_lag = lags[best_idx]
        best_corr = xcorr[best_idx] / (np.std(sig) * np.std(ask) * len(sig))

        print(f"Asset {k+1}: Best Corr = {best_corr:.4f}, Best Lag = {best_lag}")
        results.append((f"Asset {k+1}", best_corr, best_lag))

    # Pick most correlated asset
    best_asset = max(results, key=lambda x: abs(x[1]))
    print("\nMost likely match:", best_asset)


def strategy1(df, no_trade):
    signal = []
    vals = []

    # calculate MACD and signal line using exponential moving averages
    no_trade['fast_EMA']=no_trade['signal'].ewm(span=12, adjust=False).mean()
    no_trade['slow_EMA']=no_trade['signal'].ewm(span=26, adjust=False).mean()
    no_trade["MACD"]=no_trade["fast_EMA"]-no_trade["slow_EMA"]
    no_trade['Signal_Line']=no_trade['MACD'].ewm(span=9,adjust=False).mean()
    no_trade['Signal_Line']=no_trade['Signal_Line'].ewm(span=9,adjust=False).mean()
    val = 50
    for i in range(len(df)):
      if no_trade["MACD"].iloc[i]>no_trade['Signal_Line'].iloc[i]:
        signal.append(1)
        vals.append(val)
      elif no_trade['MACD'].iloc[i]<no_trade['Signal_Line'].iloc[i]:
        signal.append(-1)
        vals.append(val)
      else:
        signal.append(0)
        vals.append(0)
    df['Signal']=signal
    df['Volume']=vals
    return df


def plot_macd_lob_plotly(df, no_trade):


    fig = go.Figure()


    # MACD & Signal Line traces
    fig.add_trace(go.Scatter(x=no_trade.index, y=no_trade['MACD'], mode='lines', name='MACD', line=dict(color='blue')))
    fig.add_trace(go.Scatter(x=no_trade.index, y=no_trade['Signal_Line'], mode='lines', name='Signal Line', line=dict(color='orange')))


    # Crossover markers
    buy_signals = df[df['Signal'] == 1]
    sell_signals = df[df['Signal'] == -1]
    fig.add_trace(go.Scatter(x=buy_signals.index, y=no_trade['MACD'].loc[buy_signals.index], mode='markers', name='Bullish Crossover', marker=dict(symbol='triangle-up', color='green', size=10)))
    fig.add_trace(go.Scatter(x=sell_signals.index, y=no_trade['MACD'].loc[sell_signals.index], mode='markers', name='Bearish Crossover', marker=dict(symbol='triangle-down', color='red', size=10)))


    # LOB Prices on secondary y-axis
    ask_colors = ['#FF5733', '#FF8D1A', '#FFC300']
    bid_colors = ['#33FF57', '#1AFF8D', '#33FFF0']


    for i, col in enumerate(['ask_px1','ask_px2','ask_px3']):
        fig.add_trace(go.Scatter(x=df.index, y=df[col], mode='lines', name=col, line=dict(color=ask_colors[i]), yaxis='y2'))
    for i, col in enumerate(['bid_px1','bid_px2','bid_px3']):
        fig.add_trace(go.Scatter(x=df.index, y=df[col], mode='lines', name=col, line=dict(color=bid_colors[i]), yaxis='y2'))


    # Layout with dual y-axis
    fig.update_layout(
    title='MACD & Signal Line vs LOB Prices (Plotly)',
    xaxis=dict(title='Index'),
    yaxis=dict(title='MACD / Signal Line'),
    yaxis2=dict(title='LOB Prices', overlaying='y', side='right'),
    legend=dict(orientation='h', y=-0.2),
    template='plotly_dark'
    )


    fig.show()

#Pair_Trader
@dataclass
class SSMParams:
    theta: float
    kappa: float
    sigma: float
    gamma: float
    H: float


def _kalman_loglik_and_filter(
    y: np.ndarray,
    r_m: Optional[np.ndarray],
    params: SSMParams,
    delta: float,
    x0: float = 0.0,
    P0: float = 1.0,
):
    theta, kappa, sigma, gamma, H = (
        params.theta,
        params.kappa,
        params.sigma,
        params.gamma,
        params.H,
    )
    # filtering logic will be built on these parameters
    A = np.exp(-kappa * delta)
    Q = (sigma**2) / (2.0 * kappa) * (1.0 - np.exp(-2.0 * kappa * delta))
    n = len(y)

    x_filt = np.empty(n)
    P = P0
    x = x0

    innov = np.empty(n) # innovations (prediction errors)
    S = np.empty(n) # innovation variances
    H2 = H * H

    ll = 0.0
    for k in range(n):
        # Predict
        x_pred = theta * (1.0 - A) + A * x
        P_pred = A * P * A + Q

        mu = x_pred + (gamma * r_m[k] if r_m is not None else 0.0)
        eps = y[k] - mu

        Sk = P_pred + H2
        K = P_pred / Sk # Kalman gain

        x = x_pred + K * eps
        P = P_pred - K * P_pred

        x_filt[k] = x
        innov[k] = eps
        S[k] = Sk

        # accumulate log-likelihood contribution
        ll += -0.5 * (np.log(2 * np.pi * Sk) + (eps**2) / Sk)

    return ll, x_filt, innov, S


def _neg_loglik(params_vec, y, r_m, delta):
    # enforce positivity on scale parameters
    theta, kappa, sigma, gamma, H = params_vec
    kappa = max(kappa, 1e-6)
    sigma = max(sigma, 1e-6)
    H = max(H, 1e-6)
    params = SSMParams(theta, kappa, sigma, gamma, H)
    ll, _, _, _ = _kalman_loglik_and_filter(y, r_m, params, delta)
    return -ll #optimizer will minimize


def _initial_params(y: np.ndarray) -> SSMParams:
    # crude initial guess: mean level and residual variance
    theta_hat = float(np.nanmean(y))
    resid = y - theta_hat
    H0 = float(np.nanstd(resid, ddof=1))
    return SSMParams(theta_hat, 2.0, 0.1, 0.0, max(H0, 1e-3))


def _do_mle(y_win, r_win, start_params, delta, max_mle_iter=150):
    # parameter bounds for optimizer (keeps estimates in sensible range)
    bounds = [
        (-0.1, 0.1),   # theta
        (0.1, 20.0),   # kappa
        (0.01, 1.0),   # sigma
        (-2.0, 2.0),   # gamma
        (0.01, 1.0),   # H
    ]
    x0 = np.array(
        [start_params.theta, start_params.kappa, start_params.sigma,
         start_params.gamma, start_params.H]
    )

    # run MLE via L-BFGS-B, minimizing negative log-likelihood
    res = minimize(
        _neg_loglik, x0, args=(y_win, r_win, delta),
        method="L-BFGS-B", bounds=bounds,
        options={"maxiter": max_mle_iter, "ftol": 1e-8}
    )

    # clip results to bounds again for numerical safety
    th, kp, sg, gm, hh = res.x
    return SSMParams(
        theta=float(np.clip(th, -0.1, 0.1)),
        kappa=float(np.clip(kp, 0.1, 20.0)),
        sigma=float(np.clip(sg, 0.01, 1.0)),
        gamma=float(np.clip(gm, -2.0, 2.0)),
        H=float(np.clip(hh, 0.01, 1.0)),
    )


def stochastic_residual_spread(
    price_a: pd.Series,
    price_b: pd.Series,
    *,
    market_excess: Optional[pd.Series] = None,
    delta: float = 1 / 252,
    window_mle: int = 252,
    reestimate_every: int = 25,
    lookback_L: int = 10,
    recent_vol_lookback: int = 40,
    entry_mult: float = 1.5,
    exit_frac: float = 0.5,
) -> Tuple[pd.Series, pd.Series]:

    # compute daily returns of both assets
    Ra = (price_a).pct_change()
    Rb = (price_b).pct_change()
    df = pd.concat({"Ra": Ra, "Rb": Rb}, axis=1).dropna()
    y = (df["Ra"] - df["Rb"]).values

    if market_excess is not None:
        r_m = market_excess.reindex(df.index).values
        if np.isnan(r_m).all():
            r_m = None
    else:
        r_m = None

    params = _initial_params(y)
    n = len(y)
    x_filt = np.full(n, np.nan)
    x_prev, P_prev = 0.0, 10.0
    reest_counter = 0

    # run Kalman filter with periodic MLE re-estimation
    for k in range(n):
        if reest_counter == 0 and k >= window_mle:
            y_win = y[k - window_mle + 1:k + 1]
            r_win = None if r_m is None else r_m[k - window_mle + 1:k + 1]
            params = _do_mle(y_win, r_win, params, delta)
            reest_counter = reestimate_every
        else:
            reest_counter = max(reest_counter - 1, 0)

        # prediction step
        A = np.exp(-params.kappa * delta)
        Q = (params.sigma**2) / (2.0 * params.kappa) * (1 - np.exp(-2.0 * params.kappa * delta))
        H2 = params.H**2

        x_pred = params.theta * (1 - A) + A * x_prev
        P_pred = A * P_prev * A + Q

        # update step
        mu = x_pred + (params.gamma * (r_m[k] if r_m is not None else 0.0))
        eps = y[k] - mu
        S = P_pred + H2
        K = P_pred / S

        x_now = x_pred + K * eps
        P_now = P_pred - K * P_pred

        x_filt[k] = x_now
        x_prev, P_prev = x_now, P_now

    # rolling mean and volatility of filtered residuals
    x_ma = pd.Series(x_filt, index=df.index).rolling(lookback_L, min_periods=1).mean()
    recent_vol = x_ma.rolling(recent_vol_lookback, min_periods=10).std()
    entry_th = entry_mult * recent_vol
    exit_th = exit_frac * entry_th

    # trading position logic based on thresholds
    pos = pd.Series(0, index=df.index, dtype=int)
    theta_series = pd.Series(params.theta, index=df.index)

    for t in range(1, n):
        xa, th_enter, th_exit = x_ma.iloc[t], entry_th.iloc[t], exit_th.iloc[t]
        if not np.isfinite(xa) or not np.isfinite(th_enter):
            pos.iloc[t] = pos.iloc[t - 1]
            continue
        if xa > theta_series.iloc[t] + th_enter:
            pos.iloc[t] = -1
        elif xa < theta_series.iloc[t] - th_enter:
            pos.iloc[t] = 1
        elif abs(xa - theta_series.iloc[t]) < th_exit and pos.iloc[t - 1] != 0:
            pos.iloc[t] = 0
        else:
            pos.iloc[t] = pos.iloc[t - 1]

    sig1 = pos.copy()      # signal for asset 1
    sig2 = pos.copy()      # signal for asset 2
    sig1 = sig1.values
    sig2 = sig2.values

    return sig1, sig2


def compute_vwap_series(df: pd.DataFrame) -> pd.Series:
    vwap=(df["ask_px1"]*df["ask_sz1"]+df["ask_px2"]*df["ask_sz2"]+df["ask_px3"]*df["ask_sz3"]+df["bid_px1"]*df["bid_sz1"]+df["bid_px2"]*df["bid_sz2"]+df["bid_px3"]*df["bid_sz3"])/(df['bid_sz1']+df['bid_sz2']+df['bid_sz3']+df['ask_sz1']+df['ask_sz2']+df['ask_sz3'])
    return vwap
def Volume(df):
    df['Volume']=0

    # relative pressure: (bids - asks) / (total depth)
    df['Volume_pressure']=(df['bid_sz1']+df['bid_sz2']+df['bid_sz3']-df['ask_sz1']-df['ask_sz2']-df['ask_sz3'])/(df['bid_sz1']+df['bid_sz2']+df['bid_sz3']+df['ask_sz1']+df['ask_sz2']+df['ask_sz3'])

    # assign volume depending on imbalance strength
    for i in range(len(df)):
        if df['Volume_pressure'].iloc[i]>2/3:
          df['Volume'].iloc[i]=df['bid_sz1'].iloc[i]+df['bid_sz2'].iloc[i]
        elif df['Volume_pressure'].iloc[i]>0:
          df['Volume'].iloc[i]=df['bid_sz1'].iloc[i]
        elif df['Volume_pressure'].iloc[i]<-2/3:
          df['Volume'].iloc[i]=df['ask_sz1'].iloc[i]+df['ask_sz2'].iloc[i]
        else:
          df['Volume'].iloc[i]=df['ask_sz1'].iloc[i]
    return df

def run_stochastic_spread(asset3: pd.DataFrame, asset6: pd.DataFrame):
    # compute VWAP time series for both assets
    vwaps = [compute_vwap_series(df) for df in [asset3, asset6]]

    # generate trading signals using stochastic residual spread model
    sigs3, sigs6 = stochastic_residual_spread(vwaps[0], vwaps[1])
    sigs3 = np.insert(sigs3, 0, 0)
    sigs6 = np.insert(sigs6, 0, 0)
    asset3 = asset3.copy()
    asset6 = asset6.copy()
    asset3["Signal"] = sigs3[: len(asset3)]
    asset6["Signal"] = sigs6[: len(asset6)]
    asset3 = Volume(asset3)
    asset6 = Volume(asset6)
    asset3['Fair_Price']=vwaps[0]
    asset6['Fair_Price']=vwaps[1]

    return asset3, asset6



#  ICHIMOKU Preprocess

def collapse_trades_by_t_keep_side(trade_log: pd.DataFrame) -> pd.DataFrame:
    required = {'t', 'price', 'qty', 'side'}
    missing = required - set(trade_log.columns)
    if missing:
        raise ValueError(f"Missing required columns: {missing}")

    df = trade_log[['t', 'price', 'qty', 'side']].copy()

    # ensure numeric dtype for price/qty, fill invalid qty with 0
    df['qty'] = pd.to_numeric(df['qty'], errors='coerce').fillna(0.0)
    df['price'] = pd.to_numeric(df['price'], errors='coerce')

    side_counts = df.groupby('t')['side'].nunique(dropna=False)
    bad_ts = side_counts[side_counts > 1].index.tolist()
    if bad_ts:
        raise ValueError(f"'side' not consistent for some t values (examples): {bad_ts[:5]}")

    df['pxqty'] = df['price'] * df['qty']

    agg = df.groupby('t', sort=False, as_index=False).agg(
        pxqty=('pxqty', 'sum'),
        qty=('qty', 'sum'),
        side=('side', 'first'),
    )


    mean_price = (
        df.groupby('t', sort=False)['price']
          .mean()
          .reset_index(name='mean_price')
    )

    # merge and finalize price: VWAP if qty > 0 else mean price
    agg = agg.merge(mean_price, on='t', how='left')
    agg['price'] = np.where(agg['qty'] > 0, agg['pxqty'] / agg['qty'], agg['mean_price'])

    # final output
    out = agg[['t', 'side', 'price', 'qty']].reset_index(drop=True)
    return out

def compute_orderbook_imbalance(df, levels=3):
    df = df.copy()
    for L in range(1, levels+1):

        # take top-L bid and ask sizes
        bid_sizes = [f"bid_sz{i}" for i in range(1, L+1)]
        ask_sizes = [f"ask_sz{i}" for i in range(1, L+1)]

        # aggregate depth at each side
        bid_sum = df[bid_sizes].sum(axis=1)
        ask_sum = df[ask_sizes].sum(axis=1)

        # order book imbalance = (bids - asks) / (total depth)
        df[f"OBI_{L}"] = (bid_sum-ask_sum) / (bid_sum + ask_sum)

    return df



def signal_accuracy(df: pd.DataFrame, trade_log: pd.DataFrame, th: float = 0) -> pd.DataFrame:
    if 'OBI_3' not in df.columns:
        raise ValueError("df must contain 'OBI_3'")
    if not {'t','side'}.issubset(trade_log.columns):
        raise ValueError("trade_log must contain columns 't' and 'side'.")

    out = df.copy()

    # convert order book imbalance into trading signal:
    out['direction'] = np.where(out['OBI_3'] > th, 1,
                         np.where(out['OBI_3'] < -th, -1, 0))

    # clean up trade log: keep only time and side, normalize case/whitespace
    tl = trade_log[['t', 'side']].copy()
    tl['side'] = tl['side'].astype(str).str.strip().str.lower()
    tl = tl.drop_duplicates(subset='t', keep='first')

    try:
        if np.issubdtype(out.index.dtype, np.datetime64):
            tl['t'] = pd.to_datetime(tl['t'], errors='coerce')
        elif np.issubdtype(out.index.dtype, np.number):
            tl['t'] = pd.to_numeric(tl['t'], errors='coerce')
    except Exception:
        pass

    side_by_t = tl.set_index('t')['side']
    side_aligned = side_by_t.reindex(out.index)

    # predicted side from signal (direction)
    pred_side = pd.Series(index=out.index, dtype=object)
    pred_side[out['direction'] == 1]  = 'bid'
    pred_side[out['direction'] == -1] = 'ask'

    # overall accuracy (including neutral signals)
    correct_all = (pred_side == side_aligned).sum()
    total_all = len(out)
    acc_all = 100.0 * correct_all / total_all if total_all else 0.0

    # accuracy only on non-neutral signals and valid trades
    nz_mask = (out['direction'] != 0) & side_aligned.notna()
    denom_nz = int(nz_mask.sum())
    num_nz = int((pred_side[nz_mask] == side_aligned[nz_mask]).sum())
    acc_nz = 100.0 * num_nz / denom_nz if denom_nz else 0.0

    return out

def make_candles_simple(ob_df: pd.DataFrame,
                        trade_log: pd.DataFrame,
                        direction_col: str = "direction") -> pd.DataFrame:
    need_ob = {
        "bid_px1","bid_sz1","bid_px2","bid_sz2","bid_px3","bid_sz3",
        "ask_px1","ask_sz1","ask_px2","ask_sz2","ask_px3","ask_sz3",
        direction_col
    }
    miss = need_ob - set(ob_df.columns)
    if miss:
        raise ValueError(f"ob_df missing: {sorted(miss)}")
    if not {"t","price"}.issubset(trade_log.columns):
        raise ValueError("trade_log must have columns ['t','price'].")

    ask_pxqty = (ob_df["ask_px1"]*ob_df["ask_sz1"] +
                 ob_df["ask_px2"]*ob_df["ask_sz2"] +
                 ob_df["ask_px3"]*ob_df["ask_sz3"])
    ask_qty   = (ob_df["ask_sz1"] + ob_df["ask_sz2"] + ob_df["ask_sz3"]).replace(0, np.nan)
    high = (ask_pxqty / ask_qty).fillna(ob_df["ask_px1"])

    bid_pxqty = (ob_df["bid_px1"]*ob_df["bid_sz1"] +
                 ob_df["bid_px2"]*ob_df["bid_sz2"] +
                 ob_df["bid_px3"]*ob_df["bid_sz3"])
    bid_qty   = (ob_df["bid_sz1"] + ob_df["bid_sz2"] + ob_df["bid_sz3"]).replace(0, np.nan)
    low = (bid_pxqty / bid_qty).fillna(ob_df["bid_px1"])

    direction = ob_df[direction_col].fillna(0)
    mid = (ob_df["bid_px1"] + ob_df["ask_px1"]) / 2.0
    close = np.where(direction == 1, ob_df["bid_px1"],
             np.where(direction == -1, ob_df["ask_px1"], mid))

    tl = trade_log[["t","price"]].copy()

    try:
        if np.issubdtype(ob_df.index.dtype, np.datetime64):
            tl["t"] = pd.to_datetime(tl["t"], errors="coerce")
    except Exception:
        pass

    trade_px = tl.set_index("t")["price"].sort_index()

    open_ = trade_px.reindex(ob_df.index).ffill().shift(1)
    open_ = open_.fillna(mid)

    # require top 2 bid/ask sizes
    req_sz = ["bid_sz1","bid_sz2","ask_sz1","ask_sz2"]
    missing = [c for c in req_sz if c not in ob_df.columns]
    if missing:
        raise ValueError(f"ob_df missing size columns: {missing}")

    vol_bid = ob_df["bid_sz1"].fillna(0) + ob_df["bid_sz2"].fillna(0)
    vol_ask = ob_df["ask_sz1"].fillna(0) + ob_df["ask_sz2"].fillna(0)
    vol_mid = ob_df["bid_sz1"].fillna(0) + ob_df["ask_sz1"].fillna(0)

    # pick volume according to signal direction
    volume = pd.Series(
        np.where(direction == 1, vol_bid,
                np.where(direction == -1, vol_ask, vol_mid)),
        index=ob_df.index,
        name="Volume"
    )

    return pd.DataFrame(
        {"open": open_.values, "high": high.values, "low": low.values, "close": close, "Volume":volume},
        index=ob_df.index
    )

def fix_ohlc_bounds(candles: pd.DataFrame, *, inplace: bool = False, report: bool = False) -> pd.DataFrame:
    cols = {"open","high","low","close","Volume"}
    if not cols.issubset(candles.columns):
        missing = cols - set(candles.columns)
        raise ValueError(f"Missing columns: {missing}")

    df = candles if inplace else candles.copy()

    # Keep original high/low for reporting
    old_high = df["high"].copy()
    old_low  = df["low"].copy()

    df["high"] = pd.concat([df["high"], df["open"], df["close"]], axis=1).max(axis=1, skipna=True)
    df["low"]  = pd.concat([df["low"],  df["open"], df["close"]], axis=1).min(axis=1, skipna=True)

    if report:
        n_high = int((df["high"] != old_high).sum())
        n_low  = int((df["low"]  != old_low).sum())
        print(f"Adjusted high on {n_high} rows; low on {n_low} rows.")

    return df

def strategy(asset, df, max_pos=50, bb_window=20, bb_std=2):

    df = df.copy()
    asset = asset.copy()

    # ----------------- Ichimoku -----------------
    period9_high = df['high'].rolling(9).max()
    period9_low = df['low'].rolling(9).min()
    df['Tenkan_sen'] = (period9_high + period9_low)/2

    period26_high = df['high'].rolling(26).max()
    period26_low = df['low'].rolling(26).min()
    df['Kijun_sen'] = (period26_high + period26_low)/2

    df['Senkou_A'] = ((df['Tenkan_sen'] + df['Kijun_sen'])/2).shift(26)
    period52_high = df['high'].rolling(52).max()
    period52_low = df['low'].rolling(52).min()
    df['Senkou_B'] = ((period52_high + period52_low)/2).shift(26)

    df['Ichimoku_Signal'] = 0
    df.loc[df['Tenkan_sen'] > df['Kijun_sen'], 'Ichimoku_Signal'] = 1
    df.loc[df['Tenkan_sen'] < df['Kijun_sen'], 'Ichimoku_Signal'] = -1

    def ichimoku_strength(row):
        top = max(row.get('Senkou_A',0), row.get('Senkou_B',0))
        bottom = min(row.get('Senkou_A',0), row.get('Senkou_B',0))
        cloud_height = top - bottom + 1e-6
        if row['close'] > top:
            return min((row['close']-top)/cloud_height, 1.0)
        elif row['close'] < bottom:
            return min((bottom-row['close'])/cloud_height, 1.0)
        else:
            return min((top-row['close']+row['close']-bottom)/(2*cloud_height), 0.5)

    df['Ichimoku_Strength'] = df.apply(ichimoku_strength, axis=1)





    df['Signal'] = df['Ichimoku_Signal']
    df['Signal'] = df['Signal'].apply(np.sign)

    df['Strength'] = df['Ichimoku_Strength']


    df['Volume'] = (df['Strength']*max_pos).clip(upper=max_pos).fillna(0).astype(int)
    df['Signal'] = df['Signal'].fillna(0)

    asset['Signal'] = df['Signal']
    asset['Volume'] = df['Volume']

    return asset

def volume_fix(df,trade_log):
    df["Volume"] = 0
    df.loc[df["Signal"] != 0, "Volume"] = 30
    net_position=0
    current_pos=0
    prev_pos=0
    for i in range(2,len(df)):
      if current_pos>0:
        if df['Signal'].iloc[i]==-1:
          current_pos=0
      elif current_pos<0:
        if df['Signal'].iloc[i]==1:
          current_pos=0
      else:
        current_pos=df['Signal'].iloc[i]
      if current_pos==0 and df['Signal'].iloc[i]==0 and net_position!=0:
        if net_position>0:
          df['Signal'].iloc[i]=-1
          df['Volume'].iloc[i]=net_position
        else:
          df['Signal'].iloc[i]=1
          df['Volume'].iloc[i]=abs(net_position)
      if df['Signal'].iloc[i-1]!=0 and df['Signal'].iloc[i]==0 and trade_log['qty'].iloc[i-1]<df['Volume'].iloc[i-1]:
        if df['Signal'].iloc[i-1]>0:
          df['Signal'].iloc[i]=1
          df['Volume'].iloc[i]=df['Volume'].iloc[i-1]-trade_log['qty'].iloc[i-1]
        else:
          df['Signal'].iloc[i]=-1
          df['Volume'].iloc[i]=abs(df['Volume'].iloc[i-1]-trade_log['qty'].iloc[i-1])
      net_position+=df['Signal'].iloc[i]*min(df['Volume'].iloc[i],trade_log['qty'].iloc[i])
      prev_pos=current_pos
    return df

def runner_ichimoku(ob_df: pd.DataFrame, trade_log: pd.DataFrame, direction_col: str = "direction", price_col: str = "close") -> pd.DataFrame:
    ob_df=compute_orderbook_imbalance(ob_df)
    ob_df=signal_accuracy(ob_df,trade_log)
    filtered_log= collapse_trades_by_t_keep_side(trade_log)
    candles = make_candles_simple(ob_df, filtered_log)
    candles = fix_ohlc_bounds(candles)
    ob_df= strategy(ob_df, candles)
    ob_df["Fair_Price"]=compute_vwap_series(ob_df)
    return ob_df

import plotly.graph_objects as go

import plotly.graph_objects as go

def plot_lob_with_signals(assets, indices=[2, 4, 7]):
    colors = {1: "green", -1: "red"}

    fig = go.Figure()

    for idx, asset in enumerate(assets, start=1):  # make idx start at 1
        # Skip if not in selected indices
        if idx not in indices:
            continue

        # Skip if no signal data
        if 'Signal' not in asset.columns:
            continue

        # Plot Ask and Bid levels
        for i in range(1, 4):
            fig.add_trace(go.Scatter(
                x=asset.index, y=asset[f'ask_px{i}'],
                mode='lines',
                name=f"Asset {idx} Ask{i}",
                line=dict(width=2, dash='solid' if i==1 else 'dash' if i==2 else 'dot')
            ))
            fig.add_trace(go.Scatter(
                x=asset.index, y=asset[f'bid_px{i}'],
                mode='lines',
                name=f"Asset {idx} Bid{i}",
                line=dict(width=2, dash='solid' if i==1 else 'dash' if i==2 else 'dot')
            ))

        # Mid-price for signals
        mid_price = (asset['bid_px1'] + asset['ask_px1']) / 2

        # Plot buy/sell signals
        for sig_value, color in colors.items():
            mask = asset['Signal'] == sig_value
            fig.add_trace(go.Scatter(
                x=asset.index[mask], y=mid_price[mask],
                mode='markers',
                marker=dict(
                    color=color,
                    size=6 + asset['Volume'][mask].abs() * 0.2,  # scale marker size with volume
                    symbol='circle'
                ),
                name=f"Asset {idx} Signal {sig_value}"
            ))

    fig.update_layout(
        title="LOB Prices with Buy/Sell Signals (Assets 2, 4, 7)",
        xaxis_title="Time Steps",
        yaxis_title="Price",
        template="plotly_dark",
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1)
    )

    fig.show()



def plotter(assets):
    n = len(assets[0])  # assuming all assets have same length

    # Compute residuals
    residuals = [(assets[6]['bid_px1'].iloc[i]) - (assets[3]['ask_px3'].iloc[i]) - (assets[1]['ask_px3'].iloc[i])
                 for i in range(n)]

    # Compute correlation
    combined = (assets[3]['ask_px3'] + assets[1]['ask_px3'])
    correlation = np.corrcoef(assets[6]['bid_px1'], combined)[0, 1]

    # Create interactive Plotly figure
    fig = go.Figure()

    # Residuals line with correlation in legend
    fig.add_trace(go.Scatter(
        y=residuals,
        mode='lines',
        name=f'Residuals (Corr: {correlation:.2f})',
        line=dict(color='blue', width=2)
    ))

    # Zero reference line
    fig.add_hline(
        y=0,
        line=dict(color="black", dash="dash"),
        opacity=0.7,
        annotation_text="Zero Line",
        annotation_position="top right"
    )

    # Layout
    fig.update_layout(
        title="Residuals (U-shaped Distribution)",
        xaxis_title="Index",
        yaxis_title="Residual Value",
        template="plotly_dark"
    )

    fig.show()

#Arbitrarge Asset
def generate_signals_and_update_df(asset2, asset4, asset7, threshold=180, volume=30):

    # Extract the bid_px1 values for each asset (for the current step)
    sig7=[]
    sig2=[]
    sig4=[]
    for i in range(len(asset2)):
      bid_px1_asset2 = asset2['bid_px1'].iloc[i]
      bid_px1_asset4 = asset4['bid_px1'].iloc[i]
      bid_px1_asset7 = asset7['bid_px1'].iloc[i]

      # Calculate the difference between asset7, asset4, and asset2
      diff = bid_px1_asset7 - bid_px1_asset4 - bid_px1_asset2

      # Initialize signal for all assets to 0
      signal_2 = 0
      signal_4 = 0
      signal_7 = 0

      # Condition: if the difference is greater than the threshold (buy condition)
      if diff > threshold:
          signal_7 = -1  # Sell signal for asset7
          signal_2 = 1   # Buy signal for asset2
          signal_4 = 1   # Buy signal for asset4

      # Condition: if the difference is less than the negative threshold (sell condition)
      if diff < -threshold:
          signal_7 = 1   # Buy signal for asset7
          signal_2 = -1  # Sell signal for asset2
          signal_4 = -1  # Sell signal for asset4
      sig7.append(signal_7)
      sig2.append(signal_2)
      sig4.append(signal_4)
    # Add signal to each DataFrame
    asset2['Signal'] = sig2
    asset4['Signal'] = sig4
    asset7['Signal'] = sig7

    # Add volume column (static volume of 50 for all assets)
    asset2['Volume'] = volume
    asset4['Volume'] = volume
    asset7['Volume'] = volume

    # Return the updated DataFrames
    return asset2, asset4, asset7

def returner(asset):
    asset['Fair_Price']=compute_vwap_series(asset)
    return asset['Signal'],asset['Volume'],asset['Fair_Price']


def controller(assets,trades,notrade):
    print("---Working On Asset1 Strategy---")
    analyze_signal_correlation(assets,notrade)
    asset1=strategy1(assets[0],notrade)
    plot_macd_lob_plotly(assets[0],notrade)
    print("---Working On Asset 2, 4, 7 Arbitrarge---")
    asset2,asset4,asset7=generate_signals_and_update_df(assets[1],assets[3],assets[6])
    plotter(assets)
    plot_lob_with_signals(assets)
    print("---Working on Asset 5 Ichimoku Clouds---")
    asset5=runner_ichimoku(assets[4],trades[4])
    print("---Working on Asset 3, 6 Pair Trading---")
    asset3,asset6 = run_stochastic_spread(assets[2],assets[5])
    signal1,vol1,fair_price1=returner(asset1)
    signal2,vol2,fair_price2=returner(asset2)
    signal3,vol3,fair_price3=returner(asset3)
    signal4,vol4,fair_price4=returner(asset4)
    signal5,vol5,fair_price5=returner(asset5)
    signal6,vol6,fair_price6=returner(asset6)
    signal7,vol7,fair_price7=returner(asset7)

    volume=[vol1.shift(1),vol2.shift(1),vol3.shift(1),vol4.shift(1),vol5.shift(1),vol6.shift(1),vol7.shift(1)]
    signal=[signal1.shift(1),signal2.shift(1),signal3.shift(1),signal4.shift(1),signal5.shift(1),signal6.shift(1),signal7.shift(1)]
    fair_price=[fair_price1,fair_price2,fair_price3,fair_price4,fair_price5,fair_price6,fair_price7]

    assets = [asset1, asset2, asset3, asset4, asset5, asset6, asset7]

    return trades_new,volume,signal,fair_price





TIMESTAMPS = 15000
N_ASSETS = 7

current_positions = [0]*7 # Read Only For Algorithm - Only Utilized By Backtester
position_limits = [50]*7  # Maximum Long/Short Position Allowed Independently - Read Only Value
position_deltas =   [0]*7 # Updated By Algorithm Every Tick - Read Only For Backtester
PnL =   [0]*7 # Updated By Algorithm Every Tick - Read Only For Backtester

path_to_data = 'BOLT_Dataset_15000'

Orderbook_snapshots = []
Trade_snapshots = []

def grouping_trades(trade):

  trade['idx'] = trade.groupby('t').cumcount() + 1

  out = trade.pivot(index='t', columns='idx', values=['side','price','qty'])

  out.columns = [f"{col}{i}" for col, i in out.columns]

  out = out.reset_index()

  return out

def sort_prices_and_qty(row):
    prices = [row['price1'], row['price2'], row['price3']]
    qty = [row['qty1'], row['qty2'], row['qty3']]

    sorted_prices_indices = np.argsort(prices)

    sorted_prices = np.array(prices)[sorted_prices_indices]
    sorted_qty = np.array(qty)[sorted_prices_indices]

    row['price1'], row['price2'], row['price3'] = sorted_prices
    row['qty1'], row['qty2'], row['qty3'] = sorted_qty

    return row
trades_new=[[] for _ in range(7)]


for i in range(N_ASSETS):
    Orderbook_snapshots.append(pd.read_csv(f"https://raw.githubusercontent.com/yus11x11/Dataset_BOLT_1500/main/Asset_{i+1}_OB_snapshot_15000.csv"))
    Trade_snapshots.append(pd.read_csv(f"https://raw.githubusercontent.com/yus11x11/Dataset_BOLT_1500/main/Asset_{i+1}_TD_snapshot_15000.csv"))
notrade=pd.read_csv("https://raw.githubusercontent.com/yus11x11/Dataset_BOLT_1500/main/noTradeAsset_15000.csv")
print(Orderbook_snapshots)
print(Trade_snapshots)

for i in range(7):
  trades_new[i]=grouping_trades(Trade_snapshots[i])
  trades_new[i]=trades_new[i].fillna(0)
  trades_new[i].drop(columns=['side1', 'side2','side3'], inplace=True)
  trades_new[i]=trades_new[i].apply(sort_prices_and_qty, axis=1)

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)

trade_logs = [[] for _ in range(7)]

class Backtesting:
    def __init__(self, ob, volume, signal, fair_prices, transaction_cost=0.0001):
        self.ob = ob
        self.vol = volume
        self.signal = signal
        self.fair_prices = fair_prices
        self.transaction_cost = transaction_cost
        self.portfolio = [[] for _ in range(7)]
        self.number_of_stocks = [0] * 7   # Current positions
        self.pnl = [0] * 7                # Current PNL per asset
        self.total_pnl = 0                # Total PNL
        self.position_history = [[] for _ in range(7)]  # NEW: position history tracker

    def execute_trade(self, orderbook_prices, target_volume, signal, fair_price, asset_index):
        if signal == 1:  # Buy signal
            for j in range(3):
                ask_price = orderbook_prices[f"price{j+1}"]
                available_volume = orderbook_prices[f"qty{j+1}"]
                if target_volume > 0 and self.number_of_stocks[asset_index] < 50:
                    trade_vol = min(target_volume, available_volume, 50 - self.number_of_stocks[asset_index])

                    if trade_vol > 0:
                        self.number_of_stocks[asset_index] += trade_vol
                        self.pnl[asset_index] -= ask_price * trade_vol * (1 + self.transaction_cost)
                        target_volume -= trade_vol

                # log trade
                trade_logs[asset_index].append([
                    signal, orderbook_prices['price1'], orderbook_prices['price2'], orderbook_prices['price3'],
                    orderbook_prices['qty1'], orderbook_prices['qty2'], orderbook_prices['qty3'],
                    target_volume, fair_price, self.pnl[asset_index], self.number_of_stocks[asset_index]
                ])

        elif signal == -1:  # Sell signal
            for j in range(3):
                bid_price = orderbook_prices[f"price{3-j}"]
                available_volume = orderbook_prices[f"qty{3-j}"]

                if target_volume > 0 and self.number_of_stocks[asset_index] > -50:
                    trade_vol = min(target_volume, available_volume, abs(-50 - self.number_of_stocks[asset_index]))

                    if trade_vol > 0:
                        self.number_of_stocks[asset_index] -= trade_vol
                        self.pnl[asset_index] += bid_price * trade_vol * (1 - self.transaction_cost)
                        target_volume -= trade_vol

                # log trade
                trade_logs[asset_index].append([
                    signal, orderbook_prices['price1'], orderbook_prices['price2'], orderbook_prices['price3'],
                    orderbook_prices['qty1'], orderbook_prices['qty2'], orderbook_prices['qty3'],
                    target_volume, fair_price, self.pnl[asset_index], self.number_of_stocks[asset_index]
                ])

        # NEW: log current position snapshot
        self.position_history[asset_index].append(self.number_of_stocks[asset_index])

    def run_backtest(self):
        num_assets = len(self.ob)
        for day in range(len(self.ob[0]) - 1):
            for i in range(num_assets):
                orderbook_prices = self.ob[i].iloc[day]
                target_volume = self.vol[i].iloc[day]
                signal = self.signal[i].iloc[day]
                fair_price = self.fair_prices[i].iloc[day]
                self.portfolio[i].append(self.number_of_stocks[i]*fair_price+self.pnl[i])
                self.execute_trade(orderbook_prices, target_volume, signal, fair_price, i)

        # Final liquidation
        for i in range(num_assets):
            final_ob = self.ob[i].iloc[-1]
            if self.number_of_stocks[i] > 0:
                if final_ob["price1"] != 0:
                    self.pnl[i] += final_ob['price1'] * self.number_of_stocks[i] * (1 - self.transaction_cost)
                elif final_ob["price2"] != 0:
                    self.pnl[i] += final_ob['price2'] * self.number_of_stocks[i] * (1 - self.transaction_cost)
                else:
                    self.pnl[i] += final_ob['price3'] * self.number_of_stocks[i] * (1 - self.transaction_cost)
            elif self.number_of_stocks[i] < 0:
                self.pnl[i] -= final_ob['price3'] * abs(self.number_of_stocks[i]) * (1 + self.transaction_cost)

            self.number_of_stocks[i] = 0
            # log final flat position
            self.position_history[i].append(0)

        self.total_pnl = sum(self.pnl)
        print("Total PNL:", self.total_pnl)
        for idx in range(7):
            print(f"PNL_Asset{idx+1}:", self.pnl[idx])
        pnl_dict = {f"PNL_Asset{idx+1}": pnl_value for idx, pnl_value in enumerate(self.pnl)}
        pnl_dict["Total_PNL"] = self.total_pnl

        # Convert to DataFrame
        pnl_df = pd.DataFrame([pnl_dict])

        # Save to CSV
        pnl_df.to_csv("pnl_output.csv", index=False)

        print("PNL saved to pnl_output.csv")


        # 📈 Plot position histories
        self.plot_positions()
        self.plot_portfolio()

    def plot_positions(self):
        for i in range(7):
            fig = go.Figure()
            fig.add_trace(go.Scatter(
                y=self.position_history[i],
                mode='lines',
                name=f"Asset {i+1} Position",
                line=dict(width=2)
            ))
            fig.add_hline(
                y=0,
                line=dict(color="black", dash="dash"),
                opacity=0.7
            )
            fig.update_layout(
                title=f"Asset {i+1} Position History",
                xaxis_title="Time Steps",
                yaxis_title="Position Size",
                template="plotly_dark",
                legend=dict(x=0, y=1, bgcolor="rgba(255,255,255,0)")
            )

            fig.show()
    def plot_portfolio(self):
        for i in range(7):
            fig = go.Figure()
            fig.add_trace(go.Scatter(
                y=self.portfolio[i],
                mode='lines',
                name=f"Asset {i+1} Portfolio",
                line=dict(width=2)
            ))
            fig.add_hline(
                y=0,
                line=dict(color="black", dash="dash"),
                opacity=0.7
            )
            fig.update_layout(
                title=f"Asset {i+1} Portfolio",
                xaxis_title="Time Steps",
                yaxis_title="Position Size",
                template="plotly_dark",
                legend=dict(x=0, y=1, bgcolor="rgba(255,255,255,0)")
            )
            fig.show()


td_collapsed,volume,signal,fair_price = controller(Orderbook_snapshots,Trade_snapshots,notrade)

obj=Backtesting(td_collapsed,volume,signal,fair_price,transaction_cost=0.0001)
obj.run_backtest()
```
